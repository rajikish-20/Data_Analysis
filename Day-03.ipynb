###### Creating a new column in the dataframe

```python
# HEre, rank() function is a pandas function
# rank() function is giving rank-1 to minimum salary and higher rank value to higher salary 
df['rank'] = df['salary'].rank()
df
```

###### Creating a New Column from an Existing Column

```python
 df['rank123'] = df['rank']
```

###### Ranking Salaries and Sorting by Rank

```python
df['rank'] = df['salary'].rank()
df.sort_values('rank')
# sort_values () is used for sorting the dataframe based upon a column value
```

###### Ranking Salaries in Descending Order

```python
df['rank'] = df['salary'].rank(ascending = False)
df.sort_values('rank')
```

###### Dropping Columns from the DataFrame

```python
# Dropping the columns
# axis = 1: delete the column
# axis = 0: delete the row

df.drop(['rank', 'rank123'], axis=1)
```
