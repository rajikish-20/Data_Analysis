```python
import pandas as pd
df = pd.read_excel(r"S:\DA  notes\Excel_data.xlsx")
```

###### Selecting Rows and Columns Using iloc

```python
df.iloc[0:20, 0:5]
# 0:20 → Selects rows from position 0 to 19 (20 is excluded)
# 0:5  → Selects columns from position 0 to 4 (5 is excluded)
# iloc → Selects data using integer index positions
```

###### Selecting Rows and Columns Using iloc with Step

```python
df.iloc[0:200:10,0:5:2]
```

###### Selecting Random Columns Using iloc

```python
# iloc: Selecting specific columns using their index positions

df.iloc[0:200:10, [1, 4]]
# 0:200:10 → Selects rows from 0 to 199 with a step of 10
# [1, 4]    → Selects columns at positions 1 and 4
```

###### Selecting Specific Columns Using iloc

```python
df.iloc[1:10, [1,3,4]]
```

##### Selecting a Specific Value Using iloc

```python
# iloc to get a specific value
print(df.iloc[1, 5])
# 1 → Selects the row at index position 1
# 5 → Selects the column at index position 5
# iloc → Uses integer index positions
# print() → Displays the selected value
```

###### Selecting Age and Salary Columns Using loc

```python
# loc to get Age and Salary columns
df.loc[0:5, ["Age", "salary"]]
# 0:5 → Selects rows from index 0 to 5 (5 is included)
# ["Age", "salary"] → Selects only the Age and salary columns
# loc → Uses column names and index labels
```

###### Using Comparison Conditions in loc

```python
# Comparision condition inside the loc
df.Age > 50
# Checks which records have Age greater than 50
# Returns True or False for each row
```

###### Filtering Data Using loc with a Condition

```python
# Comparison condition inside the loc
df.loc[df.Age > 50, ["Age", "salary"]]
# Selects Age and salary of people whose age is greater than 50
```

###### Filtering Salary by Age Range

```python
# Salary of people Age > 50 and Age < 70
df.loc[(df.Age > 50) & (df.Age < 70),["Age","salary"]]
```

###### Using iloc with a Boolean Condition

```python
 # iloc does not work directly with a Pandas Series condition
# .values converts the Series into a NumPy array of True/False values
df.iloc[(df.Age > 50).values, 0:4]
```

###### Checking the Type of Boolean Array

```python
type((df.Age> 50).values)
```

###### Updating Values in a DataFrame Using a Condition

```python
###  Updating a value inside the dataframe
# replace / update all to age > 50  with 18
df[df.Age >50]
```

```python
# Replace the Age with 18 for all people whose Age is greater than 50
df.loc[df.Age > 50, "Age"] = 18
df
```

###### Updating a Specific Value in a DataFrame Using loc

```python
df.loc[0,"Name"] = "srujana"
df
```

###### Updating a Specific Value Using iloc

```python
# Update the value at row position 1 and column position 1
df.iloc[1, 1] = "Zombie"
# Display the updated DataFrame
df
```

###### Updating a Single Value Using at

```python
### at & iat for single value access
# Update the Name at row label 2
df.at[2, "Name"] = "Ramesh"

# Display the updated DataFrame
df
```

###### Updating a Single Value Using iat

```python
# iat is used to access or update a single value using integer positions
# Update the value at row position 0 and column position 1
df.iat[0, 1] = "Vijay Deverakonda"
df
```

###### Sorting Rows in a DataFrame

```python
# Sort the DataFrame by Age in ascending order
df.sort_values("Age")
```

###### Sorting Rows by Name in Descending Order

```python
# Sort the DataFrame by Name from Z to A
df.sort_values("Name", ascending=False)
```

###### Storing Sorted DataFrame in a New Variable

```python
# Sort the DataFrame by Name from Z to A and store it in df1
df1 = df.sort_values("Name", ascending=False)
# Display the sorted DataFrame
df1
```

###### Sorting the Index of a DataFrame

```python
# sort_index() is used to sort the index of the DataFrame
df1.sort_index()
```

###### Sorting the Index In-Place

```python
# Sort the index of df1 in ascending order and update df1 directly
df1.sort_index(inplace=True)
# Display the updated DataFrame
df1
```

###### Setting a Column as the DataFrame Index

```python
# Set the ID column as the index of the DataFrame
df1.set_index("ID", inplace=True)
# Display the updated DataFrame
df1
```

###### Resetting the DataFrame Index

```python
# Reset the index of the DataFrame
df1.reset_index(inplace=True)
# Display the updated DataFrame
df1
```
