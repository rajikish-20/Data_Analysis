###### Creating a New Column Using apply() with Multiple Columns

```python
# Example-1: Apply() function
# Query: Create a new column which is having salary/age
# Motive: use of apply() function on complete dataframe and access multiple columns
# Create a copy of the DataFrame
df1 = df.copy()
# Create a new column by dividing Salary by Age
df1["Salary_Per_Year_age"] = df1.apply(
    lambda row: row["salary"] / row["Age"],
    axis=1
)
df1
```

##### Data Transformation Using map()

```python
# map() performs element-wise transformation on a Pandas Series.
# It applies a function to each individual value.
# map() is mainly used with Series, not directly with the complete DataFrame.
# map() can be used with lambda functions.
# map() can also be used with a dictionary to replace/map values.
# It is useful for data cleaning, transformation, and value replacement.
```

###### Example 1: Converting Names to Uppercase Using map()

```python
# Definition:
# map() is used for element-wise transformation, similar to apply()
df2 = df.copy()
# Convert all names to uppercase
df2.Name = df2.Name.map(lambda val: str(val).upper())
df2
```

### 9. Aggregation and Grouping

###### Displaying DataFrame Information

```python
df.info()
```

###### Calculating the Total Salary

```python
# Total salary
df.salary.sum()
```

###### Calculating the Average Salary

```python
# Average salary
df.salary.mean()
```

###### Calculating the Median Salary

```python
# Median of salary
# Middle value after sorting the data
# Example: 10, 20, 30, 40, 50, 60, 70, 80

df.salary.median()
```

###### Finding the Most Frequent Profession

```python
# mode() : Most frequent data
df.Profession.mode()
```

###### Finding the Minimum Salary

```python
# Minimum salary
df.salary.min()
```

###### Counting Records in Each Category

```python
# value_counts() : Number of records related to each category
df.Profession.value_counts()
```

###### Counting Non-Null Values

```python
# count() : Count the non-null values
df.count()
```

###### Counting Null / Missing Values

```python
# Count the null (missing) values in each column
df.isnull().sum()
```

###### Grouping Data by Profession and Counting Records

```python
# groupby() : Group the data based on category
df.groupby('Profession').Name.count()
```

###### Calculating Average Salary by Profession

```python
# Find average salary based on profession
df.groupby('Profession').salary.mean()
```

###### Calculating Average Salary by Age

```python
# Find average salary based upon age
df.groupby('Age').salary.mean()
```

###### Applying Multiple Aggregate Functions Using agg()

```python
# agg() : Apply multiple aggregate functions together on data
df[["Age", "salary"]].agg(["mean", "sum"])
```

###### Applying Multiple Aggregate Functions Using agg()

```python
# Apply multiple aggregate functions
df[["Age", "salary"]].agg(["mean", "sum", "max", "min"])
```

###### Applying Different Aggregate Functions to Different Columns

```python
# Apply different aggregate functions on different columns
df.agg({
    "Age": ["min", "max"],
    "salary": ["mean", "min"]
})
```

###### Combining groupby() and agg()

```python
# groupby() + agg()
df.groupby("Profession").agg({
    "Age": ["min", "max"],
    "salary": ["mean", "min"]
})
```
