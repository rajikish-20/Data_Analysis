### Data filtering and indexing (loc,iloc,where())

###### Filtering Records Based on Salary

```python
# Query: Filter all the records where salary > 100000
df.salary > 100000
```

###### Filtering DataFrame Records Using a Condition

```python
# Query: Filter all the records where salary > 100000
df[df.salary > 100000]
```

###### Filtering Data Using Multiple Conditions

```python
# Query: Filter all the records where salary > 100000 and salary < 150000

df[(df.salary > 100000) & (df.salary < 150000)]
```

###### Filtering Records Within a Salary Range

```python
# Query: Filter all the records where salary > 100000 and salary < 150000

df[(df.salary > 100000) & (df.salary < 150000)]
```

###### Filtering Records Using OR Condition

```python
# Query: Filter all the records where salary < 100000 OR salary > 150000

df[(df.salary < 100000) | (df.salary > 150000)]
```
