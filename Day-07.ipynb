### 5. Handling of Missing Values (NaN/Null)

```python
# Def : Null/NaN vlaues are missing values. Null is different from 0.
# IF their null values in the data the nwrong analysis will happen.
# Null/NaN values represent missing data in a DataFrame.
# Null/NaN is different from 0 because 0 is a valid value.
# Missing values can affect calculations and lead to incorrect analysis.
# We should identify missing values before performing data analysis.
# Pandas provides isnull(), notnull(), fillna(), and dropna() to handle missing values.
```

```python
import pandas as pd
df = pd.read_excel(r"S:\DA  notes\Excel_data.xlsx")
```

#### Data Transformation Using apply() and map()

```python
# apply() applies a function to values in a Series or along rows/columns of a DataFrame.
# map() transforms values in a Pandas Series.
# apply() can work with custom functions and built-in functions.
# map() is commonly used for value mapping and replacement.
# Both are useful for cleaning and transforming data.
```

###### Doubling the Salary of Employees Using apply()

```python
# Query: Double the salary of the employees
df6 = df.copy()
# Double the salary of each employee
df6.Salary = df6.Salary.apply(lambda x: x * 2)
df6
```

###### Creating a Senior/Junior Column Using apply()

```python
# Query: Create a new column which divide the employees as senior and junior
# based upon the salary. if salary >300000 :senior otherwise junior
# use of apply(0 on complete dataframe
# Create a new Position column based on Salary
df6["Position"] = df6.apply(
    lambda row: "Senior" if row["Salary"] > 300000 else "Junior",
    axis=1
)
df6
```

###### Categorizing Employees as Old or Young Using apply()

```python
# Create a new category column based on Age
df6["category"] = df6.apply(
    lambda row: "Old" if row["Age"] > 40 else "Young",
    axis=1
)
df6
```
