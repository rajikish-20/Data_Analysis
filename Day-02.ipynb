##### Creating DataFrame

```python
import pandas as pd
```

```python
df = pd.DataFrame()     # it creates empty dataframe
print(df)
```

###### How to convert list to Data frame

```python
# convert list to dataframe
subjects = ["Maths", "English", "Computer", "Accounts"]

df = pd.DataFrame(subjects)
print(df)
```

###### we Have given Column name

```python
# convert list to dataframe - we have added a column 
mobile_list = ["Samsung", "iPhone", "OnePlus", "Vivo"]

df = pd.DataFrame(mobile_list, columns=["Mobile Brand"])

print(df)
```

###### We Have created a 2D array

```python
# convert list to dataframe
cities_list = ["Hyderabad", "Delhi", "Mumbai", "Chennai"]
df = pd.DataFrame(cities_list, columns=["City Name"])
df
```

###### How to add name and value to 2D Array

```python
# list of list 
books_list = [
    ["Harry Potter", 500],
    ["The Alchemist", 300],
    ["Wings of Fire", 450],
    ["Atomic Habits", 600]
]
df = pd.DataFrame(books_list, columns=["Book Name", "Price"])
df
```

###### How to add 3 columns in the 2D array

```python
grocery_list = [
    ["Rice", 5, 300],
    ["Sugar", 2, 100],
    ["Oil", 3, 450],
    ["Dal", 4, 400]
]
df = pd.DataFrame(
    grocery_list,
    columns=["Item", "Quantity", "Price"]
)
df
```

###### Converting List of Tuples to DataFrame

```python
# Converting list of tuples to DataFrame
students_list = [
    ("Ravi", 20, "B.Com", 85),
    ("Priya", 21, "B.Com", 92),
    ("Arjun", 20, "B.Com", 78),
    ("Sneha", 22, "B.Com", 88)
]
df = pd.DataFrame(
    students_list,
    columns=["Name", "Age", "Course", "Marks"]
)
df 
```

###### Converting Set of Tuples into DataFrame

```python
# Converting set of tuples into DataFrame
players_set = {
    ("Virat", 250, 12000),
    ("Rohit", 260, 11000),
    ("Rahul", 180, 7000),
    ("Hardik", 150, 4000)
}
df = pd.DataFrame(
    players_set,
    columns=["Player", "Matches", "Runs"]
)
df
```

###### Creating DataFrame from Set of Set

```python
# creating DataFrame from set of set
# we can not create the dataframe from set of set.
# because, set is mutable.
# in set, we cann't store the mutable items.
# in set, we can only store immutable items.

fruits_list = {
    {"AratiPandu",40,100},
    {"MamidiPandu",1000,200},
    {"KamalaPandu",10,1000},
    {"DrakshaPandu",300,-100}
}
df = pd.DataFrame(
    fruits_list,
    columns=["Fruits Name", "Price", "Discount"]
)
df
```

###### Creating the DataFrame from Dictionary

```python
# Creating DataFrame from Dictionary
student_data = {
    "Name": ["Shiva", "Niranjan", "Suresh", "Akash", "Lahari", "Rakesh"],
    "Salary": [40, 50, 60, 15, 25, 100]
}
df1 = pd.DataFrame(student_data)
df1
```
