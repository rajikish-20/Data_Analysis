### 1.Data Structure Creation (Series(1D Array), DataFrame(2d Array))

###### Series (1D Array)

###### How to Create Series

```python
ser1 = pd.Series()
print(ser1)
```

###### How to create series with values 

```python
# convert the list to series
list1 = [5,10,15,20,25,30]
ser2 = pd.Series(list1)
print(ser2)
print(type(ser2))
```

```python
# convert the list to series
list1 = [5,10,15,20,25,30, "Srujana"]
ser2 = pd.Series(list1)
print(ser2)
print(type(ser2))

# In pandas Strings are represented as object
# in series, all elements have the same datatype
```

###### How to convert tuple to series

```python
# convert the tuple to series
Tuple1 = [5,10,15,20,25,30]
ser2 = pd.Series(Tuple1)
print(ser2)
print(type(ser2))
```

```python
Tuple1 = [5,10,15,20,25,30]
ser2 = pd.Series(Tuple1)
print(ser2)
print(type(ser2))
print(ser2[2])
# In pands series, direct negative indexing not allowed
```

```python
uple1 = [5,10,15,20,25,30]
ser2 = pd.Series(Tuple1)
print(ser2)
print(type(ser2))
print(ser2[-2])
# In pands series, direct negative indexing not allowed
```

```python
# set cannot be converted to pandas
# set is unordered and Series is ordered, so no conversion
set1 = {5,10,15,20,25,30}
ser2 = pd.Series(set1)
print(ser2)
```

###### How to convert the numpy array into Series

```python
import numpy as np
arr1 = np.array([10,20,30,40,50])
print(arr1, type(arr1))

ser3 = pd.Series(arr1, index =['A','B','C','D','E'])
print(ser3,type(ser3))
print(ser3['A'])
```
