```python
import pandas as pd
```

### Merging Two DataFrames

```python
# Definition:
# Merge two DataFrames based on a common column.
# Functions used for combining DataFrames:
# merge()  → Combine based on a common column
# join()   → Combine based on index
# concat() → Combine DataFrames row-wise or column-wise
# Important points:
# on      → Specifies the common column used for merging
# how     → Specifies the type of join
# inner   → Returns only matching records
# left    → Returns all records from the left DataFrame
# right   → Returns all records from the right DataFrame
# outer   → Returns all records from both DataFrames
# axis=0  → Combine DataFrames row-wise
# axis=1  → Combine DataFrames column-wise
# suffixes → Handles same column names in both DataFrames
```

###### Creating Two DataFrames for merge()

```python
# merge() function

# Create the first DataFrame
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'DreamCompany': ["Google", "Nvidia", "Accenture"]
})

# Create the second DataFrame
df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'DreamPackage': [20, 24, 50]
})

# Display both DataFrames
print(df1)
print(df2)
```

###### Inner Merge Using merge()

```python
# merge() function behaves like joins in SQL
# By default, merge() performs an inner join
# Inner merge: Returns common records from both DataFrames

merge_df = pd.merge(df1, df2, on="ID")

merge_df
```

###### Adding a New Column Before Merging DataFrames

```python
# Add a new column "Role" to df1
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'DreamCompany': ["Google", "Nvidia", "Accenture"],
    'Role': ["DA", "DS", "HR"]
})

# Create the second DataFrame
df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'DreamPackage': [20, 24, 50]
})

# Display both DataFrames
print(df1)
print(df2)
```

###### Performing an Inner Merge Using merge()

```python
# Inner merge: Common records from both DataFrames
# Main point:
# inner merge returns only matching/common records
# from both DataFrames based on the common column.
merge_df = pd.merge(df1, df2, on="ID", how="inner")

merge_df
```

###### Left Merge

```python
# Left merge: Keep all records from the left DataFrame
# and matching records from the right DataFrame
# left → all records from LEFT + matching records from RIGHT
merge_df = pd.merge(df1, df2, on="ID", how="left")

merge_df
```

###### Right Merge

```python
# Right merge: Keep all records from the right DataFrame
# and matching records from the left DataFrame
# right → all records from RIGHT + matching records from LEFT
merge_df = pd.merge(df1, df2, on="ID", how="right")
merge_df
```

###### Outer Merge

```python
# Outer merge: Keep all records from both DataFrames
# Common records will appear only one time
# outer → all records from LEFT + RIGHT
# common records appear only once
merge_df = pd.merge(df1, df2, on="ID", how="outer")
merge_df
```

### join()

```python
# join() : Used to combine DataFrames mainly based on index values.
# Important points:
# join() → combines DataFrames based on index
# By default, join() performs a left join
# It can also perform inner, right, and outer joins
# on parameter → specifies the column to join on
# how parameter → specifies the type of join
# lsuffix → suffix added to overlapping columns from left DataFrame
# rsuffix → suffix added to overlapping columns from right DataFrame
# Useful when DataFrames have the same index
```

###### Creating a DataFrame with Custom Index Values

```python
df3 = pd.DataFrame(
    {
        "Name_richest": ["Ambani", "Elon", "Niranjan", "Larry"]
    },
    index=[10, 20, 30, 40]
)
df3
```

###### Creating a DataFrame with Net Worth and Custom Index

```python
df4 = pd.DataFrame(
    {
        "Networth": [10, 15, 500, 2]
    },
    index=[20, 30, 40, 50]
)
df4
```

###### Inner Join Using join()

```python
# Inner join: Keep only common index values
# NOte - # inner join → keeps only common index values
# Easy memory: INNER = COMMON ONLY
joined_df = df3.join(df4, how="inner")
joined_df
```

###### Left Join Using join()

```python
# Left join: Keep all index values from the left DataFrame
joined_df = df3.join(df4, how="left")
joined_df
```

###### Right Join Using join()

```python
# Right join: Keep all index values from the right DataFrame
joined_df = df3.join(df4, how="right")
joined_df
```

###### Outer Join Using join()

```python
# Outer join: Keep all index values from both DataFrames
joined_df = df3.join(df4, how="outer")
joined_df
```

### concat()

```python
# Definition:
# concat() is used to combine two or more DataFrames
# row-wise or column-wise.
# axis=0 → Combine rows (vertical)
# axis=1 → Combine columns (horizontal)
# axis=0 → rows are added one below another
# axis=1 → columns are added side by side
# ignore_index=True → creates a new continuous index
# By default, concat() keeps the original index values
```

###### Concatenating Multiple DataFrames Vertically

```python
df5 = pd.DataFrame({
    "ID": [1, 2],
    "Movie": ["Pappu Power", "Biryani Boss"]
})
df6 = pd.DataFrame({
    "ID": [3, 4],
    "Movie": ["Tea Glass Tiger", "Dosa Don"]
})
df7 = pd.DataFrame({
    "ID": [5, 6],
    "Movie": ["Auto Avengers", "Chai Biscuit Champion"]
})
# Concatenate DataFrames vertically
concat_df = pd.concat(
    [df5, df6, df7],
    axis=0,
    ignore_index=True
)
concat_df
```

###### Creating an Hourly Date Range Using pd.date_range()

```python
# pd.date_range() → Creates a sequence of dates and times.
# start → Starting date and time.
# end → Ending date and time.
# freq='h' → Generates a date/time value for every hour.
```

```python
# Create a date range from January 2021 to December 2026
# with an hourly frequency
date_rng = pd.date_range(
    start='2021-01-01',
    end='2026-12-31',
    freq='h'
)
print(date_rng)
```

###### Creating a Monthly Date Range Using pd.date_range()

```python
# pd.date_range() → Creates a sequence of dates.
# start → Starting date.
# end → Ending date.
# freq='ME' → Generates dates at the end of every month.
```

```python
# Create a date range from January 2021 to December 2026
# with month-end frequency
date_rng = pd.date_range(
    start='2021-01-01',
    end='2026-12-31',
    freq='ME'
)
print(date_rng)
```

###### Creating a Yearly Date Range Using pd.date_range()

```python
# pd.date_range() → Creates a sequence of dates.
# start → Starting date.
# end → Ending date.
# freq='YE' → Generates dates at the end of every year.
```

```python
# Create a date range from 2021 to 2026
# with year-end frequency
date_rng = pd.date_range(
    start='2021-01-01',
    end='2026-12-31',
    freq='YE'
)
print(date_rng)
```

###### Selecting 1000 Random Dates from a Date Range

```python
# np.random.choice() → Randomly selects values from an array.
# date_rng → Source date range from which dates are selected.
# size=1000 → Selects exactly 1000 dates.
# replace=True → Allows the same date to be selected multiple times.
# len(random_dates) → Checks the number of selected dates.
```

```python
# Pick only 1000 random records from the date range
import numpy as np
random_dates = np.random.choice(
    date_rng,
    size=1000,
    replace=True
)
print(len(random_dates))
```

###### Creating a Canteen DataFrame with Funny Names

```python
# Creating DataFrame using random_dates
canteen_df = pd.DataFrame({ 
    'timestamp': pd.to_datetime(random_dates),   
    'order_id': np.arange(1, 1001),
    # Generate order IDs from 1 to 1000    
    'food_category': np.random.choice(
        ["Samosa Simba", "Puff Pichha", "Tea Thagubothu",
         "Choki Choki Rowdy", "Bajji Bujji"],
        size=1000
    ),   
    'price': np.random.randint(10, 500, size=1000),    
    'quantity': np.random.randint(1, 5, size=1000)
})
canteen_df
```
