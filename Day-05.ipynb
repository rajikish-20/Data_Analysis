###### Filtering Data Using where()

```python
 # where() is used for filtering the data
df1 = df.copy()                    # Creates a copy of the original DataFrame
df1 = df1.where(df1.Age > 50)      # Keeps values where Age is greater than 50;
                                   # other values are replaced with NaN [Missing/empty value.]
df1                                # Displays the filtered DataFrame
```

###### Filtering Records Without NaN Values

```python
# Query: Filter records where Age is greater than 50
df[df.Age > 50]  # Returns only records where Age > 50; no NaN records are shown
```

###### Using where() to Filter Data

```python
### where(): Used for filtering data
# where() returns NaN for values that don't satisfy the condition
df1 = df.copy()  
df1                  
```

###### Using where() with a Replacement Value

```python
df1 = df.copy()  
df1 = df1.Age.where(df1.Age > 50,1) # Keeps Age if Age > 50; otherwise replaces it with 1
df1 
```

###### Indexing in Pandas Using loc and iloc

```python
# Indexing: loc & iloc
# loc  → Label-based indexing
# iloc → Integer position-based indexing
# loc & iloc is used for filtering
# these helps to select particular ROW & Column in the dataframe
```

###### loc Indexer in Pandas

```python
df.loc  # Used for label-based indexing
```

###### Selecting Rows and Columns Using loc

```python
# Query: print records 1 to 4 and columns id, name, age only.
df.loc[1:4, 'ID': 'Age'] # df.loc[start_row : end_row, col_start: col_end]
```

###### Selecting Rows and Columns Using loc

```python
# Query: Print records 10 to 40, and columns Name to Profession only.
df.loc[10:40, "Name":"Profession"]
# 10:40              → Selects rows 10 to 40 (40 is included)
# "Name":"Profession" → Selects columns from Name to Profession
# loc                 → Performs label-based indexing
```

```python
# Query: Print records 1 to 4, and columns ID, Name, Age only.
df.iloc[1:5, 0:3]
# 1:5 → Selects rows at positions 1 to 4 (5 is excluded)
# 0:3 → Selects columns at positions 0 to 2 (3 is excluded)
# iloc → Uses integer index positions
```
